### 1. Fakeddit Dataset

```python
import os
import json

# 原始数据路径
input_dir = "/data3/xupin/1_FakeNewsOnLLM/UNnamePro/TAFND/Datamodule/dataset/fakeddit/data_json"

# 获取文件列表
file_list = os.listdir(input_dir)

for file_name in file_list:
    output_file = file_name.replace(".json", "_demo.json")
    demo_items = []
    
    file_path = os.path.join(input_dir, file_name)

    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)
        
        # 取前四个 item（注意处理 list / dict）
        if isinstance(data, list):
            demo_items.extend(data[:4])
        elif isinstance(data, dict):
            # 从字典中取前4个键值对（按插入顺序）
            items = list(data.items())[:4]
            demo_items.extend([dict([item]) for item in items])
        else:
            print(f"{file_name} 不是标准的 JSON 列表或字典结构，跳过。")
    

    # 保存 demo 到新文件
    with open(output_file, "w", encoding="utf-8") as f:
        json.dump(demo_items, f, ensure_ascii=False, indent=2)

    print(f"已将前4个样本保存到 {output_file}")
```

```python
import os
import json
import shutil

# 输入 JSON 文件目录（包含 demo 文件）
input_dir = "/data3/xupin/1_FakeNewsOnLLM/UNnamePro/TAFND/Datamodule/dataset/fakeddit/data_json"

# 目标图片复制目录
output_img_dir = "/data3/xupin/1_FakeNewsOnLLM/UNnamePro/TAFND/Datamodule/dataset/fakeddit/fakeddit_images/images_all"

# 创建目标文件夹（如果不存在）
os.makedirs(output_img_dir, exist_ok=True)

# 遍历所有 demo JSON 文件
for file_name in os.listdir(input_dir):
    if not file_name.endswith("_demo.json"):
        continue  # 跳过非 demo 文件

    file_path = os.path.join(input_dir, file_name)
    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    if not isinstance(data, list):
        print(f"跳过 {file_name}：不是列表结构")
        continue

    for item in data:
        if not isinstance(item, dict):
            continue
        for key, value in item.items():
            img_path = item[key]['tweet_image_local']
            if img_path and os.path.isfile(img_path):
                try:
                    shutil.copy(img_path, output_img_dir)
                    print(f"✓ 已复制：{img_path}")
                except Exception as e:
                    print(f"复制失败：{img_path} → {e}")
            else:
                print(f"⚠️ 找不到文件：{img_path}")
```
